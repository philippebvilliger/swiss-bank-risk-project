# 01 - Get Data

This notebook is in charge of gathering the raw data available online, that will be used for the project. Financial stress data from the European Central Bank will be used alongside Swiss financial market and monetary data from the Swiss National Bank.

```python
import pandas as pd
from pathlib import Path
```

```python
cwd = Path.cwd()  

print(cwd)


raw_data_dir = cwd.joinpath("swiss-bank-risk-project", "data")
print(raw_data_dir)

raw_data_folder = raw_data_dir.joinpath("raw")
raw_data_folder.mkdir(parents=True, exist_ok=True)

print(raw_data_folder)
print(raw_data_folder.is_dir())
```

## Data sources

Here are the various data series that will be used for this project:

- the ECB New CISS index, an indicator measuring financial stress in the Euro area based on 15 different economic signals;
- the CHF/EUR exchange rate between the Swiss Franc and the Euro;
- the Swiss commercial bank bond yields;
- the Swiss government bond yields;
- the SNB policy rate.

```python
csv_files = []

for file in raw_data_folder.glob("*.csv"):
    csv_files.append(file)

if len(csv_files) != 4:
    raise FileNotFoundError("Some files are missing, you should have 4.")

print(csv_files)
print(len(csv_files))
```

```python
datasets = {}

for file in csv_files:
    datasets[file.stem] = pd.read_csv(file)

#print(datasets)

datasets.keys()
```

## Inspecting the raw data

The data sets are opened and inspected in order to get a general idea of what we are working with. We look out for attributes such as their size, column names and first observations.

This allows the user to check whether every file was loaded correctly and can be used for the project.

```python
for name, dataset in datasets.items(): # .items() goes and fetches the keys and the values
    print(name) # this is basically the key we established earlier
    print(dataset.shape)
    print(dataset.columns)
    #print(dataset.head())
    display(dataset.head(3))
```
