```python
import pandas as pd
import pathlib
from pathlib import Path
```

```python
cwd = Path.cwd()  # current working directory

print(cwd)


raw_data_dir = cwd.joinpath("swiss-bank-risk-project", "data")
print(raw_data_dir)

raw_data_folder = raw_data_dir.joinpath("raw")

processed_data_folder = raw_data_dir.joinpath("processed")
processed_data_folder.mkdir(parents=True, exist_ok=True)


print(raw_data_folder)
print(processed_data_folder)
```

```python
csv_files = []

for file in raw_data_folder.glob("*.csv"):
    csv_files.append(file)

datasets = {}

for file in csv_files:
    datasets[file.stem] = pd.read_csv(file)

print(len(csv_files))
print(datasets.keys())
```

```python
for name, df in datasets.items():
    print(name)
    print(df.shape)
    print(df.columns)
    print(df.dtypes)
    display(df.head())
```

Explain why the cells should be standardized before merging

```python
cleaned_datasets = {}

# CISS dataset

ciss = datasets["ciss_index"][["TIME_PERIOD", "OBS_VALUE"]].copy()

ciss = ciss.rename(columns={"TIME_PERIOD": "date", "OBS_VALUE": "value"})

cleaned_datasets["ciss_index"] = ciss

print("ciss_index")
print(ciss.shape)
display(ciss.head())


# CHF/EUR exchange rate

exchange_rate = datasets["chf-eur_exchange_rate_control"][["TIME_PERIOD", "OBS_VALUE"]].copy()

exchange_rate = exchange_rate.rename(columns={"TIME_PERIOD": "date", "OBS_VALUE": "value"})

cleaned_datasets["chf-eur_exchange_rate_control"] = exchange_rate

print("chf-eur_exchange_rate_control")
print(exchange_rate.shape)
display(exchange_rate.head())


# Find the SNB files

file_search = {file.stem: file for file in csv_files}

snb_rates = pd.read_csv(file_search["snb_interest_policy_rates_control"], sep=";", skiprows=2)

bond_yields = pd.read_csv(file_search["swiss_banks_and_goverment_bond_yields"], sep=";", skiprows=2)


# SNB policy rates

policy_rates = snb_rates.pivot(index="Date", columns="D0", values="Value").reset_index()

policy_rates["value"] = policy_rates["LZ"].combine_first((policy_rates["UG0"] + policy_rates["OG0"]) / 2)

policy_rates = policy_rates[["Date", "value"]].copy()

policy_rates = policy_rates.rename(columns={"Date": "date"})

cleaned_datasets["snb_interest_policy_rates_control"] = policy_rates

print("snb_interest_policy_rates_control")
print(policy_rates.shape)
display(policy_rates.head())


# Commercial bank and government bond yields

bond_data = bond_yields[bond_yields["D0"].isin(["E", "GK"])].pivot(index="Date", columns="D0", values="Value").reset_index()

bond_data = bond_data.rename(columns={"Date": "date","E": "government_bond_yield","GK": "commercial_bank_bond_yield"})

bond_data["value"] = (bond_data["commercial_bank_bond_yield"]- bond_data["government_bond_yield"])

bond_data = bond_data[["date", "commercial_bank_bond_yield", "government_bond_yield", "value"]].copy()

cleaned_datasets["commercial_bank_bond_spread"] = bond_data

print("commercial_bank_bond_spread")
print(bond_data.shape)
display(bond_data.head())

# Check that all four datasets are there

print(cleaned_datasets.keys())
print(len(cleaned_datasets))
```

```python
for name, dataframe in cleaned_datasets.items():
    time = cleaned_datasets[name]["date"]
    pd_date = pd.to_datetime(time, errors="coerce")
    dataframe["date"] = pd_date
    dataframe.sort_values("date", inplace=True)
    print(name)
    print(dataframe["date"].dtype)
    display(dataframe.head())
    
print(dataframe.isna().sum())
```

```python


for name, dataframe in cleaned_datasets.items():

    value_columns = dataframe.columns.drop("date")

    for column in value_columns:
        dataframe[column] = pd.to_numeric(dataframe[column], errors="coerce")

    print(name)
    print(dataframe.dtypes)
    print(dataframe.isna().sum())
    display(dataframe.head())
```

```python
for name, dataframe in cleaned_datasets.items():
    print(name)
    print(dataframe.isna().sum())
    print(dataframe.isna().mean() * 100)

    
```

```python
for name, dataframe in cleaned_datasets.items():
    print(name)
    if dataframe["date"].duplicated().any() :
        print("There are duplicates")
    print(dataframe["date"].duplicated().sum())
```


We explain why we are converting all 4 weekly datasets before merging them together

```python
weekly_datasets = {}

weekly_ciss = cleaned_datasets["ciss_index"].resample("W-FRI", on="date")["value"].mean().reset_index()

weekly_chf_eur_exchange_rate_control = cleaned_datasets["chf-eur_exchange_rate_control"].resample("W-FRI", on="date")["value"].mean().reset_index()

weekly_bank_bond_data = cleaned_datasets["commercial_bank_bond_spread"].resample("W-FRI", on="date")[["commercial_bank_bond_yield","government_bond_yield",
        "value"]].mean().reset_index()

weekly_policy_rate = cleaned_datasets["snb_interest_policy_rates_control"].set_index("date").resample("W-FRI").ffill().reset_index() 
# .ffill NA/NaN values by propagating the last valid observation to next valid.
# We do this because we have monthly values, given the policy rates change monthly and not daily.
# We therefore need to keep for each week until a new policy rate changes

weekly_datasets = {"ciss_index": weekly_ciss, "chf-eur_exchange_rate_control": weekly_chf_eur_exchange_rate_control, 
                   "commercial_bank_bond_spread": weekly_bank_bond_data, "snb_interest_policy_rates_control": weekly_policy_rate}

for name, weekly_df in weekly_datasets.items():
    print(name)
    print(weekly_df.shape)
    display(weekly_df.head())
```

```python
weekly_ciss.rename(columns={"value": "ciss_index"}, inplace=True)
display(weekly_datasets["ciss_index"].head())

weekly_chf_eur_exchange_rate_control.rename(columns={"value": "chf-eur_exchange_rate_control"}, inplace=True)
display(weekly_datasets["chf-eur_exchange_rate_control"].head())


weekly_bank_bond_data.rename(columns={"value": "commercial_bank_bond_spread"}, inplace=True)
display(weekly_datasets["commercial_bank_bond_spread"].head())

weekly_policy_rate.rename(columns={"value": "snb_policy_rate"}, inplace=True)
display(weekly_datasets["snb_interest_policy_rates_control"].head())
```

Just explain what you are about to do before writing the merge code

The datasets will be merged on the weekly date so that all indicators can be analysed together.

```python
clean_weekly_datasets = weekly_ciss.merge(weekly_chf_eur_exchange_rate_control, on="date", how="outer")

clean_weekly_datasets = clean_weekly_datasets.merge(weekly_bank_bond_data, on="date", how="outer")

clean_weekly_datasets = clean_weekly_datasets.merge(weekly_policy_rate, on="date", how="outer")

print(clean_weekly_datasets.shape)
display(clean_weekly_datasets.head())
display(clean_weekly_datasets.tail())
```

```python
print(clean_weekly_datasets.shape)
print(clean_weekly_datasets.isna().sum())
print(clean_weekly_datasets.isna().mean() * 100)
display(clean_weekly_datasets.head())
display(clean_weekly_datasets.tail())
```

```python
clean_weekly_datasets["snb_policy_rate"] = clean_weekly_datasets["snb_policy_rate"].ffill()

print(clean_weekly_datasets.info())
print("Duplicate dates:")
print(clean_weekly_datasets["date"].duplicated().sum())
print("Missing values:")
print(clean_weekly_datasets.isna().sum())
print("Date range:")
print(clean_weekly_datasets["date"].min())
print(clean_weekly_datasets["date"].max())
print("Shape:")
print(clean_weekly_datasets.shape)
display(clean_weekly_datasets.head())
display(clean_weekly_datasets.tail())
```

```python
output_file = processed_data_folder.joinpath("clean_weekly_data.csv")
clean_weekly_datasets.to_csv(output_file, index=False) # saves df as csv

print(output_file)
```
