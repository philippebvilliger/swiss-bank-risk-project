Here we will inspect and explore the cleaned weekly processed data we have obtained

```python
import pandas as pd
import pathlib
from pathlib import Path
import matplotlib.pyplot
```

```python

data_dir = Path.cwd() .joinpath("swiss-bank-risk-project", "data")

processed_data_folder = data_dir.joinpath("processed")
processed_data_folder.mkdir(parents=True, exist_ok=True)

clean_weekly_data = processed_data_folder.joinpath("clean_weekly_data.csv")

clean_weekly_df = pd.read_csv(clean_weekly_data, parse_dates=["date"])
# this allows us to convert the df column "date" into real date values that pandas will be able to use
# This will be used for further operations

print(processed_data_folder)
print(clean_weekly_data)
display(clean_weekly_df.head())
```

```python
print(clean_weekly_df.shape)
print(clean_weekly_df.columns)
clean_weekly_df.dtypes
print("Missing values:")
print(clean_weekly_df.isna().sum())

for column in clean_weekly_df.columns:
    missing_rows = clean_weekly_df[clean_weekly_df[column].isna()]

    print(column)
    display(missing_rows)

print (clean_weekly_df["date"].min())
print (clean_weekly_df["date"].max())
display(clean_weekly_df.head())
```

```python
clean_weekly_df.describe()
```

```python
ciss_index_plot = clean_weekly_df.plot(x="date", y="ciss_index", figsize=(10, 5), title="CISS Index Over Time",
                                       legend=False)

ciss_index_plot.set_xlabel("Date")
ciss_index_plot.set_ylabel("CISS Index")
ciss_index_plot.grid(True, alpha=0.3)

commercial_bank_bond_spread_plot = clean_weekly_df.plot(x="date", y="commercial_bank_bond_spread", figsize=(10, 5), 
                                                        title="Commercial Bank Bond Spread Over Time", legend=False)

commercial_bank_bond_spread_plot.set_xlabel("Date")
commercial_bank_bond_spread_plot.set_ylabel("Bond Spread")
commercial_bank_bond_spread_plot.grid(True, alpha=0.3)
```

```python
bond_yields_plot = clean_weekly_df.plot(x="date", y=["commercial_bank_bond_yield", "government_bond_yield"], figsize=(10, 5), 
                                        title="Commercial Bank and Government Bond Yields Over Time")

bond_yields_plot.set_xlabel("Date")
bond_yields_plot.set_ylabel("Yield %")
bond_yields_plot.grid(True, alpha=0.3)


exchange_rate_plot = clean_weekly_df.plot(x="date", y="chf-eur_exchange_rate_control", figsize=(10, 5), title="CHF/EUR Exchange Rate Over Time",
                                          legend=False)

exchange_rate_plot.set_xlabel("Date")
exchange_rate_plot.set_ylabel("CHF/EUR Exchange Rate")
exchange_rate_plot.grid(True, alpha=0.3)


policy_rate_plot = clean_weekly_df.plot(x="date", y="snb_policy_rate", figsize=(10, 5), title="SNB Policy Rate Over Time",
                                        legend=False)

policy_rate_plot.set_xlabel("Date")
policy_rate_plot.set_ylabel("Policy Rate (%)")
policy_rate_plot.grid(True, alpha=0.3)
```

```python
correlation_matrix = clean_weekly_df.drop(columns="date").corr()
# date is of no relevance in our correlation


display(correlation_matrix)
```

```python
print("Here are the top 10 highest CISS index observations:")
display(clean_weekly_df.sort_values("ciss_index",ascending=False).head(10))

print("Here are the top 10 highest commercial bank bond spread observations:")
display(clean_weekly_df.sort_values("commercial_bank_bond_spread", ascending=False).head(10))
```
